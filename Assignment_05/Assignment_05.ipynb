```python
import numpy as np

# Create a 1-D array with 5 elements
arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

# a) Add 2 to all elements
print("After adding 2:", arr + 2)

# b) Multiply all elements by 3
print("After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("After dividing by 2:", arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)

# Reverse the array
reverse_arr = arr[::-1]

print("Reversed array:", reverse_arr)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1])

# Find unique values and their frequencies
values, counts = np.unique(x, return_counts=True)

print("Values:", values)
print("Counts:", counts)

# Find the most frequent value
most_frequent = values[np.argmax(counts)]

# Find its indices
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, 1])

# Find unique values and frequencies
values, counts = np.unique(y, return_counts=True)

print("Values:", values)
print("Counts:", counts)

# Find the highest frequency
max_count = np.max(counts)

# Find all values having the highest frequency
most_frequent_values = values[counts == max_count]

print("Most frequent value(s):", most_frequent_values)
print("Frequency:", max_count)

# Find indices of each most frequent value
for value in most_frequent_values:
    indices = np.where(y == value)[0]
    print("Value", value, "occurs at indices:", indices)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Array:")
print(arr)

# a) Access 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) Access 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

```python
import numpy as np

# Create 25 evenly spaced numbers from 10 to 100
arr = np.linspace(10, 100, 25)

print("Array:")
print(arr)

# Dimensions
print("Number of dimensions:", arr.ndim)

# Shape
print("Shape:", arr.shape)

# Total number of elements
print("Total elements:", arr.size)

# Data type
print("Data type:", arr.dtype)

# Total number of bytes
print("Total bytes:", arr.nbytes)
```

```python
# Reshape the 1-D array into 25 rows and 1 column
reshaped_arr = arr.reshape(25, 1)

print("Reshaped array:")
print(reshaped_arr)

print("Shape:", reshaped_arr.shape)

# Transpose using T
transpose_arr = reshaped_arr.T

print("Transpose:")
print(transpose_arr)

print("Transpose shape:", transpose_arr.shape)
```

```python
import numpy as np

ucs420_gurleen = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_gurleen)
```

```python
# Mean
print("Mean:", np.mean(ucs420_gurleen))

# Median
print("Median:", np.median(ucs420_gurleen))

# Maximum
print("Maximum:", np.max(ucs420_gurleen))

# Minimum
print("Minimum:", np.min(ucs420_gurleen))

# Unique elements
print("Unique elements:", np.unique(ucs420_gurleen))
```

```python
reshaped_ucs420_gurleen = ucs420_gurleen.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_gurleen)

print("Shape:", reshaped_ucs420_gurleen.shape)
```

```python
resized_ucs420_gurleen = np.resize(ucs420_gurleen, (2, 3))

print("Resized array:")
print(resized_ucs420_gurleen)

print("Shape:", resized_ucs420_gurleen.shape)
```
