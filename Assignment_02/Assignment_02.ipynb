```python
# QUE 1 
roll_no = "1024170321"
L = [int(digit)* 10 for digit in roll_no]
print("Original List L: ")
print(L)
L.append(26)
print("\nAfter append(26):")
print(L)

L.insert(3,30)
print("\nAfter insert(3,30):")
print(L)

L.remove(26)
print("\nAfter remove(26):")
print(L)

removed = L.pop(2)
print("\nAfter poping:")
print(L)
print(removed)

L.sort()
print("\nAscending order:")
print(L)

L.sort(reverse=True)
print("\nDescending order:")
print(L)

print("\nFirst three elements:", L[:3])
print("\nLast three elements:", L[-3:])

average = sum(L)/len(L)
print("\nAverage is :",average)

greater_than_average = [x for x in L if x > average]
print(greater_than_average)
```

```python
# QUE 3 
import random as r
r.seed(1024170321)
numbers = [r.randint(100,900) for i in range(100)]
print(numbers)
odd_numbers = []
for s in numbers:
    if s%2 !=0:
        odd_numbers.append(s)
print("\nOdd numbers in the list : ",odd_numbers)
print("\nCount of odd numbers in the list : ",len(odd_numbers))
even_numbers = []
for s in numbers:
    if s%2 == 0:
        even_numbers.append(s)
print("\nEven numbers in the list : ",even_numbers)
print("\nCount of even numbers in the list : ",len(even_numbers))
prime_numbers = []

for s in numbers:
    prime = True

    if s < 2:
        prime = False

    for i in range(2, s):
        if s % i == 0:
            prime = False
            break

    if prime:
        prime_numbers.append(s)

print("\nPrime numbers in the list:", prime_numbers)
print("\nCount of prime numbers in the list : ",len(prime_numbers))
most_frequent = max(numbers, key=numbers.count)
print("\nThe most frequent number is : ",most_frequent)
print("\nThe frequency of the most frequent number is : ",numbers.count(most_frequent))
```

```python
# QUE 4
L = [1,0,2,4,1,7,0,3]
A = set([s*7 for s in L])
B = set([s*9 for s in L])
print("\nSet A : ",A)
print("\nSet B : ",B)
print("\nUnion of set A and B : ",A.union(B))
print("\nIntersection of set A and B : ",A.intersection(B))
print("\nDifference between set A and B : ",A.difference(B))
print("\nDifference between set B and A : ",B.difference(A))
# difference() gives elements unique to one specified set, while symmetric_difference() gives elements unique to either set combined.
print("\nSymmetric difference between set A and B : ",A.symmetric_difference(B))
print("\nIs set A a subset of B ? ",A.issubset(B))
print("\nIs set A a superset of B ? ",A.issuperset(B))
n = int(input("Enter a number : "))
A.discard(n)
print("\nUpdated A : ",A)
# discard() is safer because it does nothing if the value is absent, whereas remove() raises a KeyError.
```
