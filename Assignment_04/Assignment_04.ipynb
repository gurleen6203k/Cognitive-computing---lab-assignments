```python
import numpy as np

# Create a NumPy 1-D array with 5 elements
arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

# a) Add 2 to all elements
print("After adding 2:", arr + 2)

# b) Multiply all elements by 3
print("After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("After dividing by 2:", arr / 2)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

# Find unique values and their frequencies
values, counts = np.unique(x, return_counts=True)

# Find the most frequent value
most_frequent = values[np.argmax(counts)]

# Find indices of the most frequent value
indices = np.where(x == most_frequent)[0]

print("Array:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", np.max(counts))
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, 3])

# Find unique values and their frequencies
values, counts = np.unique(y, return_counts=True)

# Find the most frequent value
most_frequent = values[np.argmax(counts)]

# Find indices
indices = np.where(y == most_frequent)[0]

print("Array:", y)
print("Most frequent value:", most_frequent)
print("Frequency:", np.max(counts))
print("Indices:", indices)
```

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Array:")
print(arr)

# a) Access 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) Access 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

```python
import numpy as np

# Create 25 evenly spaced numbers from 10 to 100
arr = np.linspace(10, 100, 25)

print("Array:")
print(arr)

# Dimensions
print("\nDimensions:", arr.ndim)

# Shape
print("Shape:", arr.shape)

# Total number of elements
print("Total elements:", arr.size)

# Data type
print("Data type:", arr.dtype)

# Total number of bytes
print("Total bytes:", arr.nbytes)
```

```python
import numpy as np

# Create 3 x 4 array
ucs420_your_name = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_your_name)

# Mean
print("\nMean:", np.mean(ucs420_your_name))

# Median
print("Median:", np.median(ucs420_your_name))

# Maximum
print("Maximum:", np.max(ucs420_your_name))

# Minimum
print("Minimum:", np.min(ucs420_your_name))

# Unique elements
print("Unique elements:", np.unique(ucs420_your_name))
```
